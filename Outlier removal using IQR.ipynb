```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("weather_classification_data.csv")
dataset.head()
```

```python
dataset.describe()
```

```python
sns.boxplot(x="Temperature", data = dataset)
plt.show()
```

```python
dataset.shape
```

```python
q1 = dataset["Temperature"].quantile(0.25)
q1
```

```python
q3 = dataset["Temperature"].quantile(0.75)
q3
```

```python
iqr = q3-q1
iqr
```

```python
min_range = q1 - (1.5*iqr)
max_range = q3 + (1.5*iqr)
min_range , max_range
```

```python
#outlier positive tira present xa so negative minium range ko sarokar vaena
new_dataset= dataset[dataset["Temperature"]<= max_range]
new_dataset.head()
```

```python
new_dataset.shape
```

```python
plt.boxplot(x="Temperature",data=new_dataset)
plt.show()
```

```python
#lets try in Atmospheric Pressure
```

```python
plt.boxplot(x = "Atmospheric Pressure", data = dataset)
plt.show()
```

```python
Q1 = dataset["Atmospheric Pressure"].quantile(0.25)
Q3 = dataset["Atmospheric Pressure"].quantile(0.75)
IQR = Q3-Q1
Q1,Q3,IQR
```

```python
#esma tala mathi duitai thau ma outlier xa so max range ne chainxa min range ne chainxa.
Max_range = Q3 + (IQR*1.5)
Min_range = Q1- (1.5*IQR)
Max_range , Min_range
```

```python
Data_first=dataset[dataset["Atmospheric Pressure"]<=Max_range]
Data_first.head()
```

```python
plt.boxplot(x="Atmospheric Pressure",  data=Data_first)
plt.show()
```

```python
new_dataset=Data_first[Data_first["Atmospheric Pressure"]>=Min_range]
new_dataset.head()
```

```python
plt.boxplot(x="Atmospheric Pressure",data=new_dataset)
plt.show()
```

```python
new_dataset.describe()
```

```python
plt.boxplot(x="Wind Speed" , data=new_dataset)
plt.show()
```
