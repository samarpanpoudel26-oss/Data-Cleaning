Finding missing value

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv(r"C:\Users\Suchana\Desktop\pandas\File - 2022.csv")
dataset.head(3)
```

```python
dataset.shape
```

```python
#Gives us total row
dataset.shape[0]
```

```python
#helps to find the total number of null values
dataset.isnull().sum()
```

```python
#for total null value
dataset.isnull().sum().sum()
```

```python
dataset.notnull()
```

```python
#Total percentage
(dataset.isnull().sum().sum())/(dataset.shape[0]*dataset.shape[1])*100
```

```python
#for furthur efficiency,we represent it in a percentage
(dataset.isnull().sum())/(dataset.shape[0])*100
```

```python
#to find total not null value
dataset.notnull().sum().sum()
```

```python
#graphical representation of null values
sns.heatmap(dataset.isnull())
plt.show()
```
