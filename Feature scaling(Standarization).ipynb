```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Standarization
#large value ra small value ko bich ma antar hataune process lai feature scaling vaninxa ra esko euta way standarization ho.
#formula x(new)= (x(previous)-x(mean))/(standard deviation)
#final mean will be zero and variance will between  -1 and 1
```

```python
dataset = pd.read_csv("weather_classification_data.csv")
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
plt.boxplot(x="Atmospheric Pressure" , data = dataset)
plt.show()
```

```python
#we use sklearn for standard scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
ss.fit(dataset[["Atmospheric Pressure"]])
```

```python
dataset["Atmospheric Pressure__ss"]=ss.transform(dataset[["Atmospheric Pressure"]])
dataset.head()
```
