```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("File.csv")
dataset.head(6)
```

```python
dataset.isnull().sum()
```

```python
dataset.fillna(133).head(6)  #sabai null value vako thau ma esle 133 fill gardinxa.Yo ramro tarika haina
```

```python
dataset.info()
```

```python
#Backward filling.paxadi bata data fill hunxa
dataset.fillna(method="bfill",axis=1).head(6) 
#yesma hardik pandya ko strike rate na xa so tesko paxadi bata data fill hunxa.paxadi hundreds 0 xa so sr ne 0 vayo
```

```python
#Forward filling.aagadi bata data fill hunxa
dataset.fillna(method="ffill",axis=1).head(6) 
#yesma hardik pandya ko strike rate na xa so tesko agadi bata data fill hunxa.agadi bf 371.0 xa so sr ne 371.0 vayo
```

```python
dataset["SR"].mode()[0]
dataset["SR"].fillna(dataset["SR"].mode()[0],inplace=True)
dataset.head(5)
```

```python
dataset["Inns"].mode()[0]
dataset["Inns"].fillna(dataset["Inns"].mode()[0],inplace=True)
dataset.tail(10)
```

```python
for i in dataset.select_dtypes(include="object").columns:
    dataset[i].fillna(dataset[i].mode()[0],inplace=True)
```
