```python
import pandas as pd
```

```python
dict={
    "size":["s","m","l","xl","xxl","xl","m","l","s"]
}
df = pd.DataFrame(dict)
df.head()
```

```python
# Two dimensional dataset chalxa esma so double [[]] use hunxa
data_order=[["s","m","l","xl","xxl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
ol = OrdinalEncoder(categories = data_order)
df[["en_size"]]=ol.fit_transform(df[["size"]])
df
```

```python
#Using Map Function 
#It will be lot more easier
```

```python
ord_data= {"s":5,"m":6,"xl":7,"l":8,"xxl":9} #jun jati numbering gare ne hunnxa
df["en_size_map"]=df["size"].map(ord_data)
df
```

```python
import pandas as pd
```

```python
dataset = pd.read_csv("weather_classification_data.csv")
dataset.head(2)
```

```python
data_ord = [["inland","mountain","coastal"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories = data_ord)
oe.fit(dataset[["Location"]])
```

```python
dataset["en_location"] = oe.transform(dataset[["Location"]])
dataset.head()
```

```python
#Now from map function
```

```python
data_ord1 = {"Rainy":0,"Cloudy":1,"Sunny":2,"Snowy":3}
```

```python
dataset["en_Weather_Type"]=dataset["Weather Type"].map(data_ord1)
dataset.head()
                                                      
```
