```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("weather_classification_data.csv")
dataset.head(3)
```

```python
plt.boxplot(x="Atmospheric Pressure",data=dataset)
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ms = MinMaxScaler()
ms.fit(dataset[["Atmospheric Pressure"]])
```

```python
dataset["Atmospheric Pressure_ms"]=ms.transform(dataset[["Atmospheric Pressure"]])
dataset.head()
```

```python
plt.boxplot(x="Atmospheric Pressure_ms",data=dataset)
plt.show()
```
